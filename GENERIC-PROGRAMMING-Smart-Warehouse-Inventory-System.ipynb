```python
from typing import Generic, TypeVar, List

T = TypeVar('T')

class Inventory(Generic[T]):
    """Generic storage class that can hold any type T"""

    def __init__(self):
        self.items: List[T] = [] # Stores multiple items of type T

    def add_item(self, item: T) -> None:
        """Store an item of type T"""
        self.items.append(item)
        print(f"Added: {item} | Type: {type(item).__name__}")

    def get_all_items(self) -> List[T]:
        """Retrieve all stored items"""
        return self.items

    def display_inventory(self) -> None:
        """Display each item's type and value"""
        if not self.items:
            print("Inventory is empty")
            return

        print(f"\n--- Inventory Contents | Type: {T.__name__ if hasattr(T, '__name__') else T} ---")
        for i, item in enumerate(self.items, 1):
            print(f"{i}. Value: {item} | Type: {type(item).__name__}")


def main():
    print("=== SMART WAREHOUSE INVENTORY SYSTEM ===\n")

    # 1. Inventory[str] for product names
    product_names = Inventory[str]()
    product_names.add_item("iPhone 15 Pro")
    product_names.add_item("Samsung Galaxy S24")
    product_names.add_item("Dell XPS 13")
    product_names.display_inventory()

    # 2. Inventory[int] for stock quantities
    stock_quantities = Inventory[int]()
    stock_quantities.add_item(45)
    stock_quantities.add_item(120)
    stock_quantities.add_item(30)
    stock_quantities.display_inventory()

    # 3. Inventory[float] for product prices
    product_prices = Inventory[float]()
    product_prices.add_item(999.99)
    product_prices.add_item(849.50)
    product_prices.add_item(1299.00)
    product_prices.display_inventory()

    # 4. Inventory[list] for batches of serial numbers
    serial_batches = Inventory[list]()
    serial_batches.add_item(["SN001", "SN002", "SN003"])
    serial_batches.add_item(["SN101", "SN102"])
    serial_batches.display_inventory()

    # Type safety check - this would fail type checkers:
    # stock_quantities.add_item("fifty") # Expected int, got str


if __name__ == "__main__":
    main()
```
